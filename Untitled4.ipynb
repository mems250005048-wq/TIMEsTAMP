```python
def extract_date(message):
    if pd.isna(message):
        return None

    for line in message.splitlines():
        if line.startswith("Date:"):
            return line.replace("Date:", "").strip()
    return None
```

```python
df["timestamp_raw"] = df["message"].apply(extract_date)
```

```python
df["timestamp"] = pd.to_datetime(
    df["timestamp_raw"],
    errors="coerce",
    utc=True
)
```

```python
df["date"] = df["timestamp"].dt.date
df["hour"] = df["timestamp"].dt.hour
```

```python
daily_hour_counts = (
    df
    .groupby(["date", "hour"])
    .size()
    .reset_index(name="count")
)

daily_hour_counts.head()
```

```python
def analyze_one_day(day_df):
    """
    Input: data for a single day (hour, count)
    Output: start hour, end hour
    """
    active_hours = day_df[day_df["count"] > 0]["hour"]

    if len(active_hours) == 0:
        return None, None

    start_time = active_hours.min()
    end_time = active_hours.max()

    return start_time, end_time
```

```python
def find_sleep_period(day_df):
    """
    Input: data for a single day (hour, count)
    Output: sleep_start_hour, sleep_end_hour, sleep_duration
    """
    # Create full 24-hour table
    full_hours = pd.DataFrame({"hour": range(24)})
    full_day = full_hours.merge(day_df, on="hour", how="left")
    full_day["count"] = full_day["count"].fillna(0)

    # Mark inactive hours
    full_day["inactive"] = full_day["count"] == 0

    longest_sleep = 0
    sleep_start = None
    sleep_end = None

    current_start = None
    current_length = 0

    for _, row in full_day.iterrows():
        if row["inactive"]:
            if current_start is None:
                current_start = row["hour"]
            current_length += 1
        else:
            if current_length > longest_sleep:
                longest_sleep = current_length
                sleep_start = current_start
                sleep_end = row["hour"] - 1
            current_start = None
            current_length = 0

    # Final check (in case sleep goes till hour 23)
    if current_length > longest_sleep:
        longest_sleep = current_length
        sleep_start = current_start
        sleep_end = 23

    return sleep_start, sleep_end, longest_sleep
```

```python
results = []

for day in daily_hour_counts["date"].unique():
    day_data = daily_hour_counts[daily_hour_counts["date"] == day]

    start, end = analyze_one_day(day_data)
    sleep_start, sleep_end, sleep_duration = find_sleep_period(day_data)

    results.append({
        "date": day,
        "start_hour": start,
        "end_hour": end,
        "sleep_start": sleep_start,
        "sleep_end": sleep_end,
        "sleep_duration_hours": sleep_duration
    })

daily_summary = pd.DataFrame(results)
daily_summary.head()
```
